Import Depenency

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Membaca gambar dengan OpenCV

```python
img_bgr = cv2.imread('gb5.jpg') 
```

Convert ke RGB (Karena default OpenCV adalah BGR)

```python
img = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

```python
print("Shape setelah convert ke RGB:", img.shape)
```

```python
print (img)
```

```python
plt.imshow(img)
plt.title("Citra Asli")
plt.show()
```

Mendapatkan dimensi gambar

```python
height, width, _ = img.shape
```

Array kosong untuk grayscale

```python
grayscale_img = np.zeros((height, width), dtype=np.uint8)
```

Konversi RGB to Grayscale 

```python
for i in range(height):
    for j in range(width):
        r, g, b = img[i, j]
        gray = int(0.299 * r + 0.587 * g + 0.114 * b)  # Rumus luminance
        grayscale_img[i, j] = gray
```

Menampilkan Hasil Grayscale

```python
plt.imshow(grayscale_img, cmap='gray')
plt.title('Grayscale Image')
plt.show()
```
