```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# === GANTI FILE DI SINI SAJA ===
file_path = 'bunga matahari.jpg'  # <-- Ganti dengan nama file yang diinginkan
```

```python
# === Membaca Gambar ===
img = cv2.imread(file_path)
if img is None:
    print(f"❌ Gagal membuka file gambar: {file_path}")
else:
    print(f"Tipe data gambar: {img.dtype}, ukuran: {img.shape}")
```

```python
img_rgb = img[:, :, ::-1]  # BGR ke RGB
```

```python
height, width, channels = img.shape
print(f"Resolusi gambar: {width} x {height}")
```

```python
    # === Grayscale Manual ===
    print("\nKonversi ke Grayscale dimulai...")
    grayscale = [[0 for _ in range(width)] for _ in range(height)]
    KR, KG, KB = 0.299, 0.587, 0.114
    for y in range(height):
        for x in range(width):
            b = img[y, x, 0] / 255.0
            g = img[y, x, 1] / 255.0
            r = img[y, x, 2] / 255.0
            gray = KR * r + KG * g + KB * b
            grayscale[y][x] = gray
            if y % 100 == 0 and x % 100 == 0:
                print(f'Pixel[{y},{x}] -> R: {r:.3f}, G: {g:.3f}, B: {b:.3f}, Gray: {gray:.3f}')
    print("Grayscale selesai.")
```

```python
  # === Biner Manual ===
    print("\nKonversi ke Biner dimulai...")
    thresh = 0.5
    binary = [[0 for _ in range(width)] for _ in range(height)]
    for y in range(height):
        for x in range(width):
            binary[y][x] = 1 if grayscale[y][x] >= thresh else 0
            if y % 100 == 0 and x % 100 == 0:
                print(f'Pixel[{y},{x}] -> Gray: {grayscale[y][x]:.3f}, Binary: {binary[y][x]}')
    print("Biner selesai.")
```

```python
 # === Citra Terindeks Manual ===
    print("\nKonversi ke Citra Terindeks dimulai...")
    indexed = [[0 for _ in range(width)] for _ in range(height)]
    for y in range(height):
        for x in range(width):
            gray255 = int(grayscale[y][x] * 255)
            if gray255 < 64:
                indexed[y][x] = 0
            elif gray255 < 128:
                indexed[y][x] = 1
            elif gray255 < 192:
                indexed[y][x] = 2
            else:
                indexed[y][x] = 3
            if y % 100 == 0 and x % 100 == 0:
                print(f'Pixel[{y},{x}] -> Gray255: {gray255}, Indexed: {indexed[y][x]}')
    print("Terindeks selesai.")
```

```python
 # === Mapping Warna Terindeks ===
    print("\nMapping warna untuk citra terindeks...")
    colormap = np.array([
        [0, 0, 0],        # Hitam
        [85, 85, 255],    # Biru
        [85, 255, 85],    # Hijau
        [255, 85, 85]     # Merah
    ], dtype=np.uint8)

    indexed_rgb = np.zeros((height, width, 3), dtype=np.uint8)
    for y in range(height):
        for x in range(width):
            color_idx = indexed[y][x]
            indexed_rgb[y, x] = colormap[color_idx]
            if y % 100 == 0 and x % 100 == 0:
                print(f'Pixel[{y},{x}] -> Index: {color_idx}, Warna: {colormap[color_idx]}')
    print("Mapping warna selesai.")
```
